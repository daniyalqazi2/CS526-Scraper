```python
from googleapiclient.discovery import build

# Replace with your YouTube API Key
API_KEY = "AIzaSyAQdEego12Gpq1ZPab_OhN2V02MsTpi1UY" # my personal google cloud 

# Replace with the Video ID (e.g., 'dQw4w9WgXcQ' for https://www.youtube.com/watch?v=dQw4w9WgXcQ)
VIDEO_LINK = 'https://www.youtube.com/watch?v=NDsO1LT_0lw&ab_channel=MrBeast' #make sure your link has the ?v= and &ab towards the end


def get_video_details(api_key, video_link):
    idx = video_link.find('?v=')+3
    idx2 = video_link.find('&')
    video_id = video_link[idx:idx2]
    youtube = build("youtube", "v3", developerKey=api_key)
    
    # Get video details (including likes and comment count)
    video_response = youtube.videos().list(
        part="snippet,statistics",
        id=video_id
    ).execute()
    
    if not video_response["items"]:
        return "Invalid video ID or video not found."

    video_info = video_response["items"][0]
    posting_time = video_response['items'][0]['snippet']['publishedAt']
    channel_id = video_info["snippet"]["channelId"]
    channel_title = video_info["snippet"]["channelTitle"]
    like_count = video_info["statistics"].get("likeCount", "N/A")
    comment_count = video_info["statistics"].get("commentCount", "N/A")

    # Get channel details (including subscriber count)
    channel_response = youtube.channels().list(
        part="statistics",
        id=channel_id
    ).execute()

    if not channel_response["items"]:
        return "Channel not found."

    subscriber_count = channel_response["items"][0]["statistics"].get("subscriberCount", "N/A")

    return {
        "Channel Name": channel_title,
        "Subscriber Count": subscriber_count,
        "Like Count": like_count,
        "Comment Count": comment_count, 
        "Video Posted: ":posting_time,
        "Original Video Link": video_link
    }

# Get details
video_details = get_video_details(API_KEY, VIDEO_LINK)
print(video_details)

"""
Sample use with list of videos: 

video_list = [ 'https://www.youtube.com/watch?v=-AquWb_DWS8&ab_channel=DukeBasketball', 'https://www.youtube.com/watch?v=svktRVVNJO8&ab_channel=TheWallStreetJournal',
              'https://www.youtube.com/watch?v=ZKrQIWSmk60&ab_channel=CollegeBasketball', 'https://www.youtube.com/watch?v=-62v4lzVqA0&ab_channel=KTVB', 'https://www.youtube.com/watch?v=1ztGpGjO60o&ab_channel=PracticalEngineering'
]

for video in video_list: 
    video_details = get_video_details(API_KEY, video)
    print(video_details) 
"""
```
