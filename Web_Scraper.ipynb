```python
import csv
from googleapiclient.discovery import build

# Replace with your YouTube API Key
API_KEY = "AIzaSyAQdEego12Gpq1ZPab_OhN2V02MsTpi1UY"  # Your personal API key

# List of video links to process
video_list = [
    'https://www.youtube.com/watch?v=1K5oDtVAYzk&ab_channel=CGPGrey',
    'https://www.youtube.com/watch?v=lk4lBV5wgGM&ab_channel=CGPGrey',
    
]

# Function to get video details
def get_video_details(api_key, video_link):
    idx = video_link.find('?v=') + 3
    idx2 = video_link.find('&')
    video_id = video_link[idx:idx2]
    youtube = build("youtube", "v3", developerKey=api_key)
    
    # Get video details
    video_response = youtube.videos().list(
        part="snippet,statistics",
        id=video_id
    ).execute()
    
    if not video_response["items"]:
        return ["Invalid video", "N/A", "N/A", "N/A", "N/A", "N/A", "N/A", "N/A"]

    video_info = video_response["items"][0]
    posting_time = video_info['snippet']['publishedAt']
    channel_id = video_info["snippet"]["channelId"]
    channel_title = video_info["snippet"]["channelTitle"]
    like_count = video_info["statistics"].get("likeCount", "0")
    comment_count = video_info["statistics"].get("commentCount", "0")

    # Get channel details
    channel_response = youtube.channels().list(
        part="statistics",
        id=channel_id
    ).execute()

    if not channel_response["items"]:
        return ["Channel not found", "N/A", "N/A", "N/A", "N/A", "N/A", "N/A", "N/A"]

    subscriber_count = channel_response["items"][0]["statistics"].get("subscriberCount", "0")
    total_number_videos = channel_response['items'][0]['statistics']['videoCount']
    total_number_views = channel_response['items'][0]['statistics']['viewCount']

    return [
        channel_title, int(subscriber_count), int(total_number_videos), int(total_number_views),
        int(like_count), int(comment_count), posting_time, video_link
    ]

# Save data to CSV
csv_filename = "youtube_video_data.csv"

# Define CSV headers
headers = ["Channel Name", "Subscriber Count", "Total Videos", "Total Views", 
           "Like Count", "Comment Count", "Video Posted", "Video Link"]

# Open CSV file and write data
with open(csv_filename, mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerow(headers)  # Write headers

    for video in video_list:
        video_details = get_video_details(API_KEY, video)
        writer.writerow(video_details)  # Write video data

print(f"Data successfully saved to {csv_filename}")
```
